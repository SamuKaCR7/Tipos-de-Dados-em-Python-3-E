```python
# tipos de dados no python - revisão

#INT
idade = 17
print("idade :")
print(idade)
print(type(idade))

#FLOAT (Ponto Flutuante) valores com virgula
altura = 1.60
print("altura :")
print(altura)
print(type(altura))

#STR (string) texto
nome = "José Carvalho"
print("nome :")
print(nome)
print(type(nome))

#BOOL boleano SIM, NÃO, VERDADE, MENTIRA, LIGADO, DESLIGADO
briga - False
print("briga :")
print(briga)
print(type(briga))

#List Lista (Coleção Ordenada)
frutas = ["bluberry", "morango", "manga"]
print("fruta :")
print(fruta)
print(type(fruta))

#Tuple Tuplas Duplas, como por exemplo, cordenadas
cordenadas = (52.640, 24.325)
print("cordenadas :")
print(cordenadas)
print(type(cordenadas))

#Set - Conjunto
numeros = {1, 2, 3, 4, 5, 5, 5}
print("numeros :")
print(numeros)
print(type(numeros))

#Dict - Dicionario
pessoas = {"nome": "José Carvalho", "idade"}
print("pessoas :")
print(pessoas).;
print(type(pessoas))
```
